```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/images.jpeg', 0)
_, image = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

pattern1 = np.array([[1, 1, -1],
                     [1, -1, -1],
                     [-1, -1, -1]], dtype=np.int8)

pattern2 = np.array([[-1, -1, 1],
                     [-1, 1, -1],
                     [1, -1, -1]], dtype=np.int8)

result1 = cv2.morphologyEx(image, cv2.MORPH_HITMISS, pattern1)
result2 = cv2.morphologyEx(image, cv2.MORPH_HITMISS, pattern2)

cv2_imshow(image)
cv2_imshow(result1)
cv2_imshow(result2)
```
